````python
import sys
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

PROJECT_ROOT = Path("..")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.utils import (
    load_image,
    show_image,
    image_info
)
````

````python
print("=" * 60)
print("CNN FROM SCRATCH")
print("01 - IMAGE AS MATRIX")
print("=" * 60)
````

````python
# Cell 3 - Load Image

image = load_image(
    "../data/sample.jpg",
    grayscale=False
)

image_info(image)
````

````python
# Cell 4 - Display image

show_image(
    image,
    title="Original Image"
)
````

````python
# Cell 5 - understanding Shape

print("Shape =", image.shape)

channels = image.shape[0]
height = image.shape[1]
width = image.shape[2]

print()
print("Channels :", channels)
print("Height   :", height)
print("Width    :", width)
````

````python
# Cell 6 - RGB Channels

red = image[0]
green = image[1]
blue = image[2]

print(red.shape)
````

````python
# Cell 7 - Visualize RGB Channels

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(green, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.show()
````

````python
# Cell 8 - Pixel Values

row = 100
col = 100

r = image[0, row, col]
g = image[1, row, col]
b = image[2, row, col]

print("Pixel Position:")
print((row, col))

print()
print("RGB Value:")
print((r, g, b))
````

````python
# Cell 9 - Image as Matrix

print("Red Channel")

print(red[:8, :8])
````

````python
# Cell 10 - Flatten image

vector = image.reshape(-1)

print("Original Shape:")
print(image.shape)

print()

print("Flatten Shape:")
print(vector.shape)
````

````python
# Cell 11 - Number of Features

num_features = np.prod(image.shape)

print(
    "Total Features =",
    num_features
)
````

````python
# Cell 12 - Grayscale image

gray = load_image(
    "../data/sample.jpg",
    grayscale=True
)

show_image(
    gray,
    title="Grayscale"
)

print(gray.shape)
````

````python
# Cell 13 0 RGB vs Grayscale

print("RGB Features")

rgb_features = np.prod(
    image.shape
)

print(rgb_features)

print()

print("Gray Features")

gray_features = np.prod(
    gray.shape
)

print(gray_features)
````
