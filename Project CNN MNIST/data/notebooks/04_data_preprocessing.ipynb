````python
from pathlib import Path
import json

import numpy as np

from torchvision.datasets import MNIST
````

````python
ROOT_DIR = Path("../")

RAW_DIR = ROOT_DIR / "raw"
REPORT_DIR = ROOT_DIR / "reports"
INTERIM_DIR = ROOT_DIR / "interim"

INTERIM_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print(ROOT_DIR.resolve())
print(RAW_DIR.resolve())
print(REPORT_DIR.resolve())
print(INTERIM_DIR.resolve())
````

````python
validation_path = (
    REPORT_DIR /
    "validation.json"
)

with open(
    validation_path,
    "r"
) as f:

    validation = json.load(f)

validation
````

````python
assert (
    validation["overall_status"] is True
), "Dataset validation failed."

print(
    "Validation passed."
)
````

````python
train_dataset = MNIST(
    root=RAW_DIR / "mnist",
    train=True,
    download=False
)

test_dataset = MNIST(
    root=RAW_DIR / "mnist",
    train=False,
    download=False
)
````

````python
X_train = train_dataset.data.numpy()

y_train = train_dataset.targets.numpy()

X_test = test_dataset.data.numpy()

y_test = test_dataset.targets.numpy()
````

````python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
````

````python
print(
    "Train Min:",
    X_train.min()
)

print(
    "Train Max:",
    X_train.max()
)

print(
    "Test Min:",
    X_test.min()
)

print(
    "Test Max:",
    X_test.max()
)
````

````python
X_train = X_train.astype(
    np.float32
)

X_test = X_test.astype(
    np.float32
)
````

````python
X_train = X_train / 255.0

X_test = X_test / 255.0
````

````python
print(
    "Train Range:",
    X_train.min(),
    X_train.max()
)

print(
    "Test Range:",
    X_test.min(),
    X_test.max()
)
````

````python
print(
    "X_train dtype:",
    X_train.dtype
)

print(
    "X_test dtype:",
    X_test.dtype
)
````

````python
preprocessing_metadata = {

    "input_dtype": "uint8",

    "output_dtype": "float32",

    "normalization": "divide_by_255",

    "pixel_range_before": [
        0,
        255
    ],

    "pixel_range_after": [
        0.0,
        1.0
    ],

    "train_shape": list(
        X_train.shape
    ),

    "test_shape": list(
        X_test.shape
    )
}
````

````python
with open(
    INTERIM_DIR /
    "preprocessing_metadata.json",
    "w"
) as f:

    json.dump(
        preprocessing_metadata,
        f,
        indent=4
    )
````

````python
np.save(
    INTERIM_DIR /
    "X_train_normalized.npy",
    X_train
)

np.save(
    INTERIM_DIR /
    "y_train.npy",
    y_train
)

np.save(
    INTERIM_DIR /
    "X_test_normalized.npy",
    X_test
)

np.save(
    INTERIM_DIR /
    "y_test.npy",
    y_test
)
````
