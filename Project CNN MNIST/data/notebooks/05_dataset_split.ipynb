````python
from pathlib import Path
import json

import numpy as np

from sklearn.model_selection import train_test_split
````

````python
ROOT_DIR = Path("../")

INTERIM_DIR = ROOT_DIR / "interim"

PROCESSED_DIR = ROOT_DIR / "processed"

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)
````

````python
RANDOM_STATE = 42

VALID_SIZE = 0.1
````

````python
X_train_full = np.load(
    INTERIM_DIR /
    "X_train_normalized.npy"
)

y_train_full = np.load(
    INTERIM_DIR /
    "y_train.npy"
)

X_test = np.load(
    INTERIM_DIR /
    "X_test_normalized.npy"
)

y_test = np.load(
    INTERIM_DIR /
    "y_test.npy"
)
````

````python
print(X_train_full.shape)
print(y_train_full.shape)

print(X_test.shape)
print(y_test.shape)
````

````python
X_train, X_valid, y_train, y_valid = train_test_split(
    X_train_full,
    y_train_full,
    test_size=VALID_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_train_full
)
````

````python
print()

print("Train:")
print(X_train.shape)
print(y_train.shape)

print()

print("Valid:")
print(X_valid.shape)
print(y_valid.shape)

print()

print("Test:")
print(X_test.shape)
print(y_test.shape)
````

````python
split_metadata = {

    "random_state": RANDOM_STATE,

    "validation_size": VALID_SIZE,

    "train_samples": int(
        len(X_train)
    ),

    "validation_samples": int(
        len(X_valid)
    ),

    "test_samples": int(
        len(X_test)
    ),

    "train_shape": list(
        X_train.shape
    ),

    "validation_shape": list(
        X_valid.shape
    ),

    "test_shape": list(
        X_test.shape
    )
}
````

````python
with open(
    PROCESSED_DIR /
    "split_metadata.json",
    "w"
) as f:

    json.dump(
        split_metadata,
        f,
        indent=4
    )
````

````python
np.save(
    PROCESSED_DIR /
    "X_train.npy",
    X_train
)

np.save(
    PROCESSED_DIR /
    "y_train.npy",
    y_train
)

np.save(
    PROCESSED_DIR /
    "X_valid.npy",
    X_valid
)

np.save(
    PROCESSED_DIR /
    "y_valid.npy",
    y_valid
)

np.save(
    PROCESSED_DIR /
    "X_test.npy",
    X_test
)

np.save(
    PROCESSED_DIR /
    "y_test.npy",
    y_test
)
````
