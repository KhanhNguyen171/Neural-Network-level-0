mục tiêu:

`Download Dataset -> Save to raw/`

Input: `Internet`

Output: `data/raw/mnist/`

```python
from torchvision.datasets import MNIST
```

```python
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()

RAW_DIR = PROJECT_ROOT / "raw"
MNIST_DIR = RAW_DIR / "mnist"

RAW_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MNIST_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print(f"PROJECT_ROOT : {PROJECT_ROOT}")
print(f"RAW_DIR      : {RAW_DIR}")
print(f"MNIST_DIR    : {MNIST_DIR}")
```

```python
train_dataset = MNIST(
    root=str(MNIST_DIR),
    train=True,
    download=True
)

print(
    f"Training samples: {len(train_dataset)}"
)
```

```python
test_dataset = MNIST(
    root=str(MNIST_DIR),
    train=False,
    download=True
)

print(
    f"Test samples: {len(test_dataset)}"
)
```

```python
for path in sorted(MNIST_DIR.rglob("*")):
    print(path.relative_to(PROJECT_ROOT))
```

```python
dataset_info = {
    "dataset_name": "MNIST",
    "train_samples": len(train_dataset),
    "test_samples": len(test_dataset),
    "image_height": 28,
    "image_width": 28,
    "channels": 1,
    "num_classes": 10
}

for key, value in dataset_info.items():
    print(f"{key:20}: {value}")
```

```python
import json

manifest = {
    "dataset": "MNIST",
    "status": "downloaded",
    "train_samples": len(train_dataset),
    "test_samples": len(test_dataset)
}
```

```python
manifest_path = RAW_DIR / "ingestion_manifest.json"

with open(
    manifest_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        indent=4
    )

print(
    f"Saved: {manifest_path}"
)
```
