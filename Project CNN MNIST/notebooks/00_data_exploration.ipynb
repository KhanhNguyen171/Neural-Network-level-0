# Data Exploration Notebook

**Purpose**: Explore and visualize the MNIST dataset structure, distribution, and characteristics

**Outputs**: Dataset statistics, distribution plots, sample visualizations

## 1. Import Libraries and Load Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Load data
data_dir = Path('data/processed')
X_train = np.load(data_dir / 'X_train.npy')
X_valid = np.load(data_dir / 'X_valid.npy')
X_test = np.load(data_dir / 'X_test.npy')
y_train = np.load(data_dir / 'y_train.npy')
y_valid = np.load(data_dir / 'y_valid.npy')
y_test = np.load(data_dir / 'y_test.npy')

print("✓ Data loaded successfully")
print(f"Training set: {X_train.shape}, Labels: {y_train.shape}")
print(f"Validation set: {X_valid.shape}, Labels: {y_valid.shape}")
print(f"Test set: {X_test.shape}, Labels: {y_test.shape}")
```

## 2. Dataset Statistics

```python
# Dataset statistics
total_samples = len(X_train) + len(X_valid) + len(X_test)
stats_data = {
    'Split': ['Training', 'Validation', 'Test', 'Total'],
    'Samples': [len(X_train), len(X_valid), len(X_test), total_samples],
    'Percentage': [
        f"{len(X_train)/total_samples*100:.1f}%",
        f"{len(X_valid)/total_samples*100:.1f}%",
        f"{len(X_test)/total_samples*100:.1f}%",
        "100%"
    ]
}

stats_df = pd.DataFrame(stats_data)
print("\n📊 Dataset Split:")
print(stats_df.to_string(index=False))

# Pixel statistics
print(f"\n📈 Pixel Value Statistics (Training):")
print(f"  Min: {X_train.min():.4f}")
print(f"  Max: {X_train.max():.4f}")
print(f"  Mean: {X_train.mean():.4f}")
print(f"  Std: {X_train.std():.4f}")
print(f"  Normalized Range: [0, 1]" if X_train.max() <= 1 else "  Normalized Range: [0, 255]")
```

## 3. Class Distribution

```python
# Class distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for idx, (y, title) in enumerate([(y_train, 'Training'), (y_valid, 'Validation'), (y_test, 'Test')]):
    unique, counts = np.unique(y, return_counts=True)
    axes[idx].bar(unique, counts, color='steelblue', alpha=0.7, edgecolor='black')
    axes[idx].set_xlabel('Digit', fontsize=11)
    axes[idx].set_ylabel('Count', fontsize=11)
    axes[idx].set_title(f'{title} Set Distribution', fontsize=12, fontweight='bold')
    axes[idx].grid(True, alpha=0.3, axis='y')
    axes[idx].set_xticks(range(10))

plt.tight_layout()
plt.savefig('notebooks/assets/figures/01_class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Class distribution visualization saved")
```

## 4. Sample Visualizations

```python
# Visualize sample digits - one per class
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
axes = axes.flatten()

for digit in range(10):
    idx = np.where(y_train == digit)[0][0]
    ax = axes[digit]
    ax.imshow(X_train[idx], cmap='gray')
    ax.set_title(f'Digit {digit}', fontsize=11, fontweight='bold')
    ax.axis('off')

plt.suptitle('Sample MNIST Digits (One per Class)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('notebooks/assets/figures/02_sample_digits.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Sample digits visualization saved")
```

## 5. Data Quality Checks

```python
# Quality checks
print("\n✓ Data Quality Checks:")
print(f"  No NaN values in X_train: {not np.isnan(X_train).any()}")
print(f"  No NaN values in y_train: {not np.isnan(y_train).any()}")
print(f"  All pixel values normalized [0,1]: {X_train.min() >= 0 and X_train.max() <= 1}")
print(f"  Labels in range [0,9]: {y_train.min() == 0 and y_train.max() == 9}")
print(f"  Image shape (28x28): {X_train.shape[1:] == (28, 28)}")

# Data type check
print(f"\n  X_train dtype: {X_train.dtype}")
print(f"  y_train dtype: {y_train.dtype}")
print(f"  Data size: {X_train.nbytes / 1024 / 1024:.2f} MB")

print("\n✅ All quality checks passed!")
```

## 6. Summary Statistics

```python
# Create summary
summary_stats = {
    'Metric': [
        'Total Samples',
        'Training Samples',
        'Validation Samples',
        'Test Samples',
        'Image Size',
        'Number of Classes',
        'Pixel Min',
        'Pixel Max',
        'Pixel Mean',
        'Data Format'
    ],
    'Value': [
        f"{total_samples:,}",
        f"{len(X_train):,}",
        f"{len(X_valid):,}",
        f"{len(X_test):,}",
        '28x28',
        '10 (0-9)',
        f"{X_train.min():.4f}",
        f"{X_train.max():.4f}",
        f"{X_train.mean():.4f}",
        'NumPy (.npy)'
    ]
}

summary_df = pd.DataFrame(summary_stats)
print("\n" + "="*50)
print("📊 Data Exploration Summary")
print("="*50)
print(summary_df.to_string(index=False))
print("="*50)

# Save summary
summary_df.to_csv('notebooks/assets/tables/00_data_summary.csv', index=False)
print("\n✓ Summary saved to CSV")
```

# Data Exploration Notebook

**Purpose**: Explore and visualize the MNIST dataset structure, distribution, and characteristics

## 1. Load Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Load data
data_dir = Path('data/processed')
X_train = np.load(data_dir / 'X_train.npy')
X_valid = np.load(data_dir / 'X_valid.npy')
X_test = np.load(data_dir / 'X_test.npy')
y_train = np.load(data_dir / 'y_train.npy')
y_valid = np.load(data_dir / 'y_valid.npy')
y_test = np.load(data_dir / 'y_test.npy')

print("✓ Data loaded successfully")
print(f"Training set: {X_train.shape}")
print(f"Validation set: {X_valid.shape}")
print(f"Test set: {X_test.shape}")
```

## 2. Dataset Statistics

```python
# Dataset statistics
stats_data = {
    'Split': ['Training', 'Validation', 'Test', 'Total'],
    'Samples': [len(X_train), len(X_valid), len(X_test), len(X_train) + len(X_valid) + len(X_test)],
    'Percentage': [
        f"{len(X_train)/(len(X_train) + len(X_valid) + len(X_test))*100:.1f}%",
        f"{len(X_valid)/(len(X_train) + len(X_valid) + len(X_test))*100:.1f}%",
        f"{len(X_test)/(len(X_train) + len(X_valid) + len(X_test))*100:.1f}%",
        "100%"
    ]
}

stats_df = pd.DataFrame(stats_data)
print("\n📊 Dataset Split:")
print(stats_df.to_string(index=False))

# Pixel statistics
print(f"\n📈 Pixel Value Statistics:")
print(f"  Min: {X_train.min():.4f}")
print(f"  Max: {X_train.max():.4f}")
print(f"  Mean: {X_train.mean():.4f}")
print(f"  Std: {X_train.std():.4f}")
```

## 3. Class Distribution

```python
# Class distribution
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for idx, (y, title) in enumerate([(y_train, 'Training'), (y_valid, 'Validation'), (y_test, 'Test')]):
    unique, counts = np.unique(y, return_counts=True)
    axes[idx].bar(unique, counts, color='steelblue', alpha=0.7)
    axes[idx].set_xlabel('Digit')
    axes[idx].set_ylabel('Count')
    axes[idx].set_title(f'{title} Set Distribution')
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('notebooks/assets/figures/class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Class distribution saved")
```

## 4. Sample Visualizations

```python
# Visualize sample digits
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for digit in range(10):
    idx = np.where(y_train == digit)[0][0]
    ax = axes[digit // 5, digit % 5]
    ax.imshow(X_train[idx], cmap='gray')
    ax.set_title(f'Digit {digit}')
    ax.axis('off')

plt.suptitle('Sample MNIST Digits', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('notebooks/assets/figures/sample_digits.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Sample digits saved")
```
