## 1. Load Training Results

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
import json

# Load checkpoint
checkpoint_path = Path('artifacts/checkpoints/last.pt')
if checkpoint_path.exists():
    checkpoint = torch.load(checkpoint_path, map_location='cpu')
    print(f"✓ Checkpoint loaded: {checkpoint_path}")
    print(f"  Epoch: {checkpoint.get('epoch', 'unknown')}")
    print(f"  Keys: {list(checkpoint.keys())}")
else:
    print("⚠ No checkpoint found. Run training first:")
    print("  python -m scripts.train --config configs/experiments/baseline.yaml")
```

## 2. Load Logs

```python
# Check for training logs
log_dir = Path('artifacts/logs')
if log_dir.exists():
    log_files = list(log_dir.glob('*.json'))
    print(f"✓ Found {len(log_files)} log files:")
    for f in log_files[:5]:  # Show first 5
        print(f"  - {f.name}")
else:
    print("⚠ No logs found. Logs will be created during training.")
```
