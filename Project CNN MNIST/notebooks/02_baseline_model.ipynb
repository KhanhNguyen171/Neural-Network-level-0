# Baseline Model Training

**Purpose**: Train CNN baseline model on MNIST dataset

**Outputs**: Trained model checkpoint, training metrics, accuracy plots

## 1. Setup and Configuration

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys
import yaml

sys.path.append('.')
from src.models.factory import ModelFactory
from src.data.dataloader import create_train_dataloader, create_test_dataloader

print(f"PyTorch version: {torch.__version__}")
print(f"GPU Available: {torch.cuda.is_available()}")
print(f"Device: {torch.device('cuda' if torch.cuda.is_available() else 'cpu')}")

# Load config
config_path = 'configs/experiments/baseline.yaml'
with open(config_path) as f:
    config = yaml.safe_load(f)

print(f"\n✓ Config loaded from {config_path}")
print(f"  Epochs: {config['training']['num_epochs']}")
print(f"  Batch size: {config['data']['batch_size']}")
print(f"  Learning rate: {config['optimizer']['lr']}")
```

## 2. Create Model

```python
# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Create model
model_factory = ModelFactory()
model = model_factory.create(
    name=config['model']['name'],
    num_classes=config['model']['num_classes'],
    hidden_features=config['model']['hidden_features']
)

model = model.to(device)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"✓ Model created: {config['model']['name']}")
print(f"  Total parameters: {total_params:,}")
print(f"  Trainable parameters: {trainable_params:,}")
print(f"  Device: {device}")
```

## 3. Load Data

```python
# Create data loaders
train_loader = create_train_dataloader(
    data_dir=config['data']['data_dir'],
    batch_size=config['data']['batch_size']
)

val_loader = create_test_dataloader(
    data_dir=config['data']['data_dir'],
    batch_size=config['data']['batch_size'],
    split='valid'
)

print(f"✓ Data loaders created")
print(f"  Training batches: {len(train_loader)}")
print(f"  Validation batches: {len(val_loader)}")

# Show batch
x_batch, y_batch = next(iter(train_loader))
print(f"  Batch shape: {x_batch.shape}")
print(f"  Labels: {y_batch.unique().tolist()}")
```
