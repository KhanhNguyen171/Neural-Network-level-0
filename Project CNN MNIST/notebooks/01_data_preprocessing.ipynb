# Data Preprocessing Notebook

**Purpose**: Prepare data through normalization and augmentation techniques

**Outputs**: Preprocessed statistics, augmentation visualizations, validation reports

## 1. Load Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from scipy import ndimage
from skimage.transform import rotate

# Load data
data_dir = Path('data/processed')
X_train = np.load(data_dir / 'X_train.npy')
y_train = np.load(data_dir / 'y_train.npy')

print("✓ Data loaded successfully")
print(f"Shape: {X_train.shape}")
print(f"Range: [{X_train.min():.4f}, {X_train.max():.4f}]")
```

## 2. Data Normalization Check

```python
# Verify normalization
print("\n📊 Normalization Statistics:")
print(f"  Min value: {X_train.min():.6f}")
print(f"  Max value: {X_train.max():.6f}")
print(f"  Mean: {X_train.mean():.6f}")
print(f"  Std Dev: {X_train.std():.6f}")
print(f"  Median: {np.median(X_train):.6f}")

# Percentiles
percentiles = [1, 5, 25, 50, 75, 95, 99]
print(f"\n  Percentiles:")
for p in percentiles:
    val = np.percentile(X_train, p)
    print(f"    {p}th: {val:.6f}")

if X_train.min() >= 0 and X_train.max() <= 1:
    print("\n✅ Data is properly normalized to [0, 1]")
else:
    print("\n⚠️  Data needs normalization")
```

## 3. Data Augmentation Examples

```python
# Data augmentation techniques
def augment_image(img, angle=15, shift=2):
    """Apply augmentation: rotation and shift"""
    # Rotation
    rotated = rotate(img, angle=angle, preserve_range=True)
    
    # Shift
    shifted = ndimage.shift(rotated, shift=[shift, shift], cval=0)
    
    return rotated, shifted

# Select sample digit and visualize augmentations
sample_idx = np.where(y_train == 7)[0][0]
sample_img = X_train[sample_idx]

# Generate augmentations
rotated, shifted = augment_image(sample_img)

# Visualize
fig, axes = plt.subplots(1, 4, figsize=(12, 3))

axes[0].imshow(sample_img, cmap='gray')
axes[0].set_title('Original', fontweight='bold')
axes[0].axis('off')

axes[1].imshow(rotated, cmap='gray')
axes[1].set_title('Rotation (+15°)', fontweight='bold')
axes[1].axis('off')

axes[2].imshow(shifted, cmap='gray')
axes[2].set_title('Shift (2px)', fontweight='bold')
axes[2].axis('off')

# Gaussian noise
noisy = sample_img + np.random.normal(0, 0.05, sample_img.shape)
noisy = np.clip(noisy, 0, 1)
axes[3].imshow(noisy, cmap='gray')
axes[3].set_title('Gaussian Noise', fontweight='bold')
axes[3].axis('off')

plt.suptitle('Data Augmentation Techniques', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('notebooks/assets/figures/03_augmentation_examples.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Augmentation visualization saved")
```

# Data Preprocessing Notebook

**Purpose**: Prepare data for model training with preprocessing and augmentation

**Outputs**: Normalized data, augmentation samples, preprocessing metrics

## 1. Load and Inspect Data

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys

sys.path.append('.')
from src.data.validation import validate_data
from src.data.dataset import MNISTDataset

# Load data
data_dir = Path('data/processed')
X_train = np.load(data_dir / 'X_train.npy')
y_train = np.load(data_dir / 'y_train.npy')

print(f"✓ Data loaded: {X_train.shape}")
print(f"  Min: {X_train.min():.4f}, Max: {X_train.max():.4f}")
print(f"  Mean: {X_train.mean():.4f}, Std: {X_train.std():.4f}")
```

## 2. Normalization

```python
# Check if already normalized
if X_train.min() >= 0 and X_train.max() <= 1:
    print("✓ Data already normalized to [0, 1]")
else:
    print("⚠ Normalizing data to [0, 1]")
    X_train = X_train / 255.0

print(f"\nAfter normalization:")
print(f"  Min: {X_train.min():.4f}")
print(f"  Max: {X_train.max():.4f}")
print(f"  Mean: {X_train.mean():.4f}")
print(f"  Std: {X_train.std():.4f}")
```

## 3. Data Augmentation

```python
# Example augmentation techniques for MNIST
from scipy.ndimage import rotate, shift

def augment_image(img, rotation_range=15, shift_range=2):
    """Apply random augmentation to image"""
    # Random rotation
    angle = np.random.uniform(-rotation_range, rotation_range)
    aug_img = rotate(img, angle, reshape=False)
    
    # Random shift
    shift_val = np.random.uniform(-shift_range, shift_range, 2)
    aug_img = shift(aug_img, shift_val)
    
    return np.clip(aug_img, 0, 1)

# Show augmentation examples
fig, axes = plt.subplots(3, 5, figsize=(12, 8))

sample_idx = np.random.randint(0, len(X_train))
original = X_train[sample_idx]

# Original
axes[0, 0].imshow(original, cmap='gray')
axes[0, 0].set_title('Original')
axes[0, 0].axis('off')

# Augmented versions
for i in range(1, 5):
    aug = augment_image(original)
    axes[0, i].imshow(aug, cmap='gray')
    axes[0, i].set_title(f'Aug {i}')
    axes[0, i].axis('off')

# Repeat for 2 more samples
for row in range(1, 3):
    sample_idx = np.random.randint(0, len(X_train))
    original = X_train[sample_idx]
    axes[row, 0].imshow(original, cmap='gray')
    axes[row, 0].set_title('Original')
    axes[row, 0].axis('off')
    
    for i in range(1, 5):
        aug = augment_image(original)
        axes[row, i].imshow(aug, cmap='gray')
        axes[row, i].axis('off')

plt.suptitle('Data Augmentation Examples', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('notebooks/assets/figures/data_augmentation.png', dpi=150, bbox_inches='tight')
plt.show()

print("✓ Augmentation examples saved")
```

## 4. Dataset Statistics After Processing
